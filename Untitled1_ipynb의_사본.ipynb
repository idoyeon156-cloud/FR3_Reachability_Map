```python
import sys
import subprocess
from pathlib import Path

REPO = Path("/content/sampled_reachability_maps")
REPO_COMMIT = "cd49112daebac83e4f66ab145b8c2aa78a1477d9"

if not REPO.exists():
    subprocess.run([
        "git", "clone",
        "https://github.com/pearl-robot-lab/sampled_reachability_maps.git",
        str(REPO)
    ], check=True)

subprocess.run([
    "git", "-C", str(REPO), "checkout", REPO_COMMIT
], check=True)

subprocess.run([
    sys.executable, "-m", "pip", "install", "-q",
    "-e", str(REPO),
    "lxml", "pyyaml", "six", "h5py", "matplotlib", "plotly", "tqdm"
], check=True)

# PyPI의 다른 버전보다 이 레포지토리 코드를 우선 사용
sys.path.insert(0, str(REPO))

import torch
import numpy as np
import pytorch_kinematics as pk

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
DTYPE = torch.float32

print("Device:", DEVICE)
print("Kinematics source:", pk.__file__)
```

```python
import urllib.request
import xml.etree.ElementTree as ET

PIPER_COMMIT = "ac41fcbcdda598f01b51cf6175ed9a24d0dacadc"
URDF_URL = (
    "https://raw.githubusercontent.com/agilexrobotics/piper_ros/"
    f"{PIPER_COMMIT}/src/piper_description/urdf/"
    "piper_description.urdf"
)

URDF_PATH = Path("/content/piper_description.urdf")
urllib.request.urlretrieve(URDF_URL, URDF_PATH)

robot_xml = ET.parse(URDF_PATH).getroot()

print("Available links:")
print([link.attrib["name"] for link in robot_xml.findall("link")])

print("\nMovable joints:")
for joint in robot_xml.findall("joint"):
    if joint.attrib["type"] != "fixed":
        limit = joint.find("limit")
        print(
            joint.attrib["name"],
            joint.attrib["type"],
            None if limit is None else limit.attrib
        )
```

```python
BASE_LINK = "base_link"
FLANGE_LINK = "link6"

# link6 좌표계에서 표현한 TCP 위치 [m], 회전 [rad]
# 기본값: link6 원점을 그대로 사용
TCP_XYZ = [0.0, 0.0, 0.0]
TCP_RPY = [0.0, 0.0, 0.0]

EE_LINK = "reachability_tcp"

root = ET.fromstring(URDF_PATH.read_text())

link_names = {x.attrib["name"] for x in root.findall("link")}
assert BASE_LINK in link_names
assert FLANGE_LINK in link_names
assert EE_LINK not in link_names

ET.SubElement(root, "link", name=EE_LINK)

tcp_joint = ET.SubElement(
    root, "joint",
    name="reachability_tcp_fixed_joint",
    type="fixed"
)
ET.SubElement(tcp_joint, "parent", link=FLANGE_LINK)
ET.SubElement(tcp_joint, "child", link=EE_LINK)
ET.SubElement(
    tcp_joint, "origin",
    xyz=" ".join(map(str, TCP_XYZ)),
    rpy=" ".join(map(str, TCP_RPY))
)

MODEL_URDF = ET.tostring(root, encoding="unicode")

chain = pk.build_serial_chain_from_urdf(
    MODEL_URDF,
    EE_LINK,
    BASE_LINK
).to(dtype=DTYPE, device=DEVICE)

joint_names = chain.get_joint_parameter_names()
joint_xml = {j.attrib["name"]: j for j in root.findall("joint")}

lower, upper = [], []

for name in joint_names:
    joint = joint_xml[name]

    # 이번 코드는 PiPER 팔의 회전관절만 대상으로 함
    assert joint.attrib["type"] == "revolute", (
        f"{name}: 별도 범위 설정이 필요한 관절입니다."
    )
    assert joint.find("mimic") is None

    limit = joint.find("limit")
    assert limit is not None, f"{name}: 관절 한계가 없습니다."

    lower.append(float(limit.attrib["lower"]))
    upper.append(float(limit.attrib["upper"]))

lower = np.asarray(lower, dtype=np.float32)
upper = np.asarray(upper, dtype=np.float32)

assert len(joint_names) == 6, joint_names
assert np.all(upper > lower)

q_min = torch.tensor(lower, dtype=DTYPE, device=DEVICE)
q_max = torch.tensor(upper, dtype=DTYPE, device=DEVICE)

print("Base:", BASE_LINK)
print("End effector:", EE_LINK)
print("TCP offset:", TCP_XYZ, TCP_RPY)

for name, lo, hi in zip(joint_names, lower, upper):
    print(f"{name}: {np.degrees(lo):.1f} ~ {np.degrees(hi):.1f} deg")

q_mid = ((q_min + q_max) / 2).unsqueeze(0)

with torch.no_grad():
    T_mid = chain.forward_kinematics(q_mid).get_matrix()

print("\nMid-range TCP position [m]:")
print(T_mid[0, :3, 3].cpu().numpy())
```

```python
@torch.no_grad()
def check_jacobian():
    q = q_mid.clone()
    eps = 1e-3

    T0 = chain.forward_kinematics(q).get_matrix()
    R0 = T0[:, :3, :3]

    J = chain.jacobian(q)
    J_fd = torch.zeros_like(J)

    for k in range(len(joint_names)):
        qp, qm = q.clone(), q.clone()
        qp[:, k] += eps
        qm[:, k] -= eps

        Tp = chain.forward_kinematics(qp).get_matrix()
        Tm = chain.forward_kinematics(qm).get_matrix()

        J_fd[:, :3, k] = (
            Tp[:, :3, 3] - Tm[:, :3, 3]
        ) / (2 * eps)

        dR = (Tp[:, :3, :3] - Tm[:, :3, :3]) / (2 * eps)
        W = dR @ R0.transpose(1, 2)
        W = (W - W.transpose(1, 2)) / 2

        J_fd[:, 3:, k] = torch.stack([
            W[:, 2, 1],
            W[:, 0, 2],
            W[:, 1, 0]
        ], dim=1)

    error = (J - J_fd).abs().max().item()
    print("Maximum Jacobian error:", error)

    assert torch.allclose(
        J, J_fd, atol=2e-3, rtol=1e-2
    ), "Jacobian 검증 실패: 모델과 라이브러리를 확인하세요."

check_jacobian()
```

```python
SEED = 42
N_SAMPLES = 1_000_000
BATCH_SIZE = 10_000
RESOLUTION = 0.02  # m

# base_link 기준의 계산 범위
GRID_MIN = np.array([-1.0, -1.0, -1.0])
GRID_MAX = np.array([ 1.0,  1.0,  1.0])

SHAPE = tuple(
    np.ceil((GRID_MAX - GRID_MIN) / RESOLUTION).astype(int)
)
GRID_MAX_EFFECTIVE = GRID_MIN + np.array(SHAPE) * RESOLUTION

N_VOXELS = int(np.prod(SHAPE))

counts_flat = np.zeros(N_VOXELS, dtype=np.int64)
manip_flat = np.zeros(N_VOXELS, dtype=np.float32)

torch.manual_seed(SEED)
if DEVICE == "cuda":
    torch.cuda.manual_seed_all(SEED)

print("Grid shape:", SHAPE)
print("Number of voxels:", N_VOXELS)
print(
    "Map arrays [MB]:",
    (counts_flat.nbytes + manip_flat.nbytes) / 1e6
)
```

```python
from tqdm.auto import tqdm

out_of_bounds = 0

with torch.no_grad():
    for start in tqdm(range(0, N_SAMPLES, BATCH_SIZE)):
        n = min(BATCH_SIZE, N_SAMPLES - start)

        q = q_min + torch.rand(
            n, len(joint_names),
            dtype=DTYPE, device=DEVICE
        ) * (q_max - q_min)

        T = chain.forward_kinematics(q).get_matrix()
        xyz = T[:, :3, 3].cpu().numpy()

        J = chain.jacobian(q)

        # sqrt(det(J J^T)) = 特異値の積
        manipulability = (
            torch.linalg.svdvals(J).prod(dim=1).cpu().numpy()
        )

        if (
            not np.isfinite(xyz).all()
            or not np.isfinite(manipulability).all()
        ):
            raise RuntimeError("FK/Jacobian에 비정상 수치가 있습니다.")

        ijk = np.floor(
            (xyz - GRID_MIN) / RESOLUTION
        ).astype(np.int64)

        inside = (
            (ijk >= 0).all(axis=1)
            & (ijk < np.array(SHAPE)).all(axis=1)
        )

        out_of_bounds += int((~inside).sum())

        flat = np.ravel_multi_index(ijk[inside].T, SHAPE)

        # 같은 배치에서 동일 격자를 여러 번 방문해도 정확히 집계
        np.add.at(counts_flat, flat, 1)
        np.maximum.at(
            manip_flat, flat, manipulability[inside]
        )

assert out_of_bounds == 0, (
    f"{out_of_bounds}개 샘플이 계산 범위를 벗어났습니다. "
    "GRID_MIN/MAX를 확장하고 5~6번 셀을 다시 실행하세요."
)
assert int(counts_flat.sum()) == N_SAMPLES

counts = counts_flat.reshape(SHAPE)
max_manip = manip_flat.reshape(SHAPE)
observed = counts > 0

occupied_ijk = np.argwhere(observed)
centers = GRID_MIN + (occupied_ijk + 0.5) * RESOLUTION

occupied_counts = counts[observed]
occupied_manip = max_manip[observed]

print("Sampled configurations:", int(counts.sum()))
print("Observed voxels:", len(centers))
print("Minimum observed position [m]:", centers.min(axis=0))
print("Maximum observed position [m]:", centers.max(axis=0))
```

```python
import json
import h5py
import plotly.graph_objects as go

OUT = Path("/content/piper_reachability")
OUT.mkdir(parents=True, exist_ok=True)

metadata = {
    "map_type": "sampled_position_reachability",
    "reference_repository_commit": REPO_COMMIT,
    "piper_urdf_commit": PIPER_COMMIT,
    "base_link": BASE_LINK,
    "flange_link": FLANGE_LINK,
    "tcp_xyz_m": TCP_XYZ,
    "tcp_rpy_rad": TCP_RPY,
    "joint_names": joint_names,
    "joint_lower_rad": lower.tolist(),
    "joint_upper_rad": upper.tolist(),
    "seed": SEED,
    "n_samples": N_SAMPLES,
    "resolution_m": RESOLUTION,
    "grid_min_m": GRID_MIN.tolist(),
    "grid_max_effective_m": GRID_MAX_EFFECTIVE.tolist(),
    "shape_xyz": [int(v) for v in SHAPE],
    "collision_checked": False,
    "orientation_constrained": False,
    "manipulability": "product_of_singular_values_of_6D_Jacobian"
}

(OUT / "metadata.json").write_text(
    json.dumps(metadata, indent=2), encoding="utf-8"
)
(OUT / "model_used.urdf").write_text(
    MODEL_URDF, encoding="utf-8"
)

np.savez_compressed(
    OUT / "map_3d.npz",
    counts=counts,
    observed=observed,
    max_manipulability=max_manip,
    grid_min=GRID_MIN,
    resolution=RESOLUTION
)

np.savetxt(
    OUT / "map_3d.csv",
    np.column_stack([
        centers, occupied_counts, occupied_manip
    ]),
    delimiter=",",
    header="x_m,y_m,z_m,visits,max_manipulability",
    comments=""
)

# 独自HDF5形式。原レポジトリのROS可視化形式とは異なる
with h5py.File(OUT / "map_3d.h5", "w") as f:
    f.create_dataset("xyz_m", data=centers, compression="gzip")
    f.create_dataset(
        "visits", data=occupied_counts, compression="gzip"
    )
    f.create_dataset(
        "max_manipulability",
        data=occupied_manip, compression="gzip"
    )
    f.attrs["resolution_m"] = RESOLUTION
    f.attrs["base_link"] = BASE_LINK
    f.attrs["metadata_json"] = json.dumps(metadata)

fig3d = go.Figure(go.Scatter3d(
    x=centers[:, 0],
    y=centers[:, 1],
    z=centers[:, 2],
    mode="markers",
    marker=dict(
        size=2,
        color=occupied_manip,
        colorscale="Viridis",
        colorbar=dict(title="Max manipulability"),
        opacity=0.7
    )
))

fig3d.update_layout(
    title=f"PiPER sampled reachability — {BASE_LINK}",
    scene=dict(
        xaxis_title="X [m]",
        yaxis_title="Y [m]",
        zaxis_title="Z [m]",
        aspectmode="data"
    ),
    height=700
)

fig3d.write_html(
    str(OUT / "map_3d.html"),
    include_plotlyjs=True
)
fig3d.show()
```

```python
import matplotlib.pyplot as plt

projection_specs = [
    ("XY", 2, 0, 1),
    ("XZ", 1, 0, 2),
    ("YZ", 0, 1, 2),
]
axis_labels = ["X", "Y", "Z"]

fig, axes = plt.subplots(3, 3, figsize=(16, 14))

for col, (name, collapse_axis, a, b) in enumerate(projection_specs):
    visits_2d = counts.sum(axis=collapse_axis)
    observed_2d = visits_2d > 0
    manip_2d = max_manip.max(axis=collapse_axis)

    assert int(visits_2d.sum()) == N_SAMPLES

    np.savez_compressed(
        OUT / f"projection_{name.lower()}.npz",
        observed=observed_2d,
        visits=visits_2d,
        max_manipulability=manip_2d,
        axes=np.array([a, b]),
        grid_min=GRID_MIN[[a, b]],
        resolution=RESOLUTION
    )

    extent = [
        GRID_MIN[a], GRID_MAX_EFFECTIVE[a],
        GRID_MIN[b], GRID_MAX_EFFECTIVE[b]
    ]

    panels = [
        (
            observed_2d.astype(float),
            "Observed region: 1 / not observed: 0",
            "Greys"
        ),
        (
            np.log1p(visits_2d),
            "log(1 + projected visits)",
            "viridis"
        ),
        (
            np.ma.masked_where(~observed_2d, manip_2d),
            "Projected max manipulability",
            "viridis"
        )
    ]

    for row, (values, label, cmap) in enumerate(panels):
        ax = axes[row, col]

        image = ax.imshow(
            values.T,
            origin="lower",
            extent=extent,
            interpolation="nearest",
            aspect="equal",
            cmap=cmap
        )

        ax.set_title(f"{name}: {label}")
        ax.set_xlabel(f"{axis_labels[a]} [m]")
        ax.set_ylabel(f"{axis_labels[b]} [m]")
        fig.colorbar(image, ax=ax, shrink=0.8)

fig.tight_layout()
fig.savefig(
    OUT / "projections_xy_xz_yz.png",
    dpi=200, bbox_inches="tight"
)
plt.show()
```
